### Data Cleaning & Analysis

```python
# Setup cell
plt.rcParams['figure.dpi'] = 50
plt.rcParams['figure.figsize'] = (8, 5)
sns.set_theme(style = "whitegrid")

df['Attrition_Num'] = (df['Attrition'] == 'Yes').astype(int)
print("Data loaded:", df.shape)
```

```python
# These columns have only 1 unique value — useless for analysis
cols_to_drop = ['EmployeeCount', 'Over18', 'StandardHours']
df = df.drop(columns=cols_to_drop)
print("Dropped 3 useless columns")
print("New shape:", df.shape)
```

```python
cate_cols = ['Attrition', 'BusinessTravel', 'Department', 'EducationField', 'Gender', 'JobRole', 'MaritalStatus', 'OverTime']

for col in cate_cols:
    print(f"{col}: {df[col].unique()}")
```

```python
# Attrition rate by department
dept_attrition = df.groupby('Department')['Attrition_Num'].mean() * 100
dept_attrition = dept_attrition.sort_values(ascending = False).round(2)
print("Attrition rate by Department (%):")
print(dept_attrition)
```

```python
# Salary analysis by attrition
salary_attrition = df.groupby('Attrition')['MonthlyIncome'].describe()
print("Salary stats by Attrition:")
print(salary_attrition)
```

```python
# Overtime analysis
overtime_attrition = df.groupby('OverTime')['Attrition_Num'].mean() * 100
print("Attrition rate by Overtime (%):")
print(overtime_attrition)
```

```python
# Age analysis -- Creating age groups
df['AgeGroup'] = pd.cut(df['Age'],
                       bins = [18, 25, 35, 45, 55, 65],
                       labels = ['18-25', '26-35', '36-45', '46-55', '56-65'])
age_attrition = df.groupby('AgeGroup')['Attrition_Num'].mean() * 100
print("Attrition rate by Age Group (%):")
print(age_attrition)
```

```python
# Job Satisfaction Analysis
satisfaction_attrition = df.groupby('JobSatisfaction')['Attrition_Num'].mean() * 100
print("Attrition rate by Job Satisfaction (1 = Low, 4 = High):")
print(satisfaction_attrition)

worklife_attrition = df.groupby('WorkLifeBalance')['Attrition_Num'].mean() * 100
print("\nAttrition rate by Work Life Balance (1 = Low, 4 = Best):")
print(worklife_attrition)
```
